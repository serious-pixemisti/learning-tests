```python
# 2x + 3y = 8
# 5x - y = 9

A = np.array([
    [2, 3],
    [5, -1]
])
b = np.array([8, 9])

x = np.linalg.solve(A, b)
print(f"Solution: x = {x[0]:.2f}, y = {x[1]:.2f}")
print(A @ x)
```

```python
# 3x + 2y + z = 10
# 2x + 3y + 2z = 15
# x + 2y + 3z = 14

A = np.array([
    [3, 2, 1],
    [2, 3, 2],
    [1, 2, 3]
])
b = np.array([10, 15, 14])
x = np.linalg.solve(A, b)
print(x)
print(np.allclose(A @ x, b))
```

```python
A = np.array([
    [4, 7],
    [2, 6]
])
A_inv = np.linalg.inv(A)
print(A_inv)
print(A @ A_inv)
print(np.allclose(A @ A_inv, np.identity(2)))
```

```python
A = np.array([
    [1, 2, 3],
    [2, 4, 6],
    [1, 1, 1]
])
print(np.linalg.det(A))
print(np.linalg.matrix_rank(A))

# Rank is 2 because row 2 is a scalar multiple of row 1 and so isn't independent
```

```python
data = np.array([
    [10, 20, 30],
    [12, 18, 33],
    [ 9, 21, 28],
    [15, 25, 40],
    [11, 19, 31]
])
# Columns = variables, Rows = experiments

d_mean = data.mean(axis=0)
print(f"Variable Means: {d_mean}")

d_std = data.std(axis=0)
print(f"Variable STDs: {d_std}")

z = (data - d_mean) / d_std
print(f"Standardized Data:\n{z}")

fvg11 = data[data[:, 0] > 11]
print(f"Experiments where first variable is greater than 11:\n{fvg11}")

norms = np.linalg.norm(data, axis=1)
print(f"Experiment Euclidean Norms: {norms}")

trans = data.T
print(f"Transposed Data:\n{trans}")
```
